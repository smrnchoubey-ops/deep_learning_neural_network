```python
import numpy as np

def initialize_parameters(layer_dims):
    np.random.seed(3)
    
    parameters = {}
    L = len(layer_dims)

    for l in range(1, L):
        parameters['W' + str(l)] = np.ones(
            (layer_dims[l], layer_dims[l-1])
        ) * 0.1
        
        parameters['b' + str(l)] = np.zeros(
            (layer_dims[l], 1)
        )

    return parameters


parameters = initialize_parameters([2, 2, 1])

print(parameters)
```

```python
def linear_forward(A_prev, W, b):
    
    Z = np.dot(W, A_prev) + b
    
    return Z
```

```python
#forward propagation for L layers

def L_layer_forward(X, parameters):
    
    A = X
    L = len(parameters) // 2   # no. of layers in the neural network
    
    for l in range(1, L + 1):
        A_prev = A
        
        W = parameters['W' + str(l)]
        b = parameters['b' + str(l)]

        #print("A" + str(l-1) + ": ", A_prev)
        #print("W" + str(l) + ": ", W)
        #print("b" + str(l) + ": ", b)

        Z = linear_forward(A_prev, W, b)
        #print("Z" + str(l) + ": ", Z)
        #print("**" * 20)

        A = Z

    return A, A_prev
```

```python
def update_parameters(parameters, y, y_hat, A1, X):

    # Output layer
    parameters['W2'][0][0] = parameters['W2'][0][0] + (
        0.001 * 2 * (y - y_hat) * A1[0][0]
    )

    parameters['W2'][0][1] = parameters['W2'][0][1] + (
        0.001 * 2 * (y - y_hat) * A1[1][0]
    )

    parameters['b2'][0][0] = parameters['b2'][0][0] + (
        0.001 * 2 * (y - y_hat)
    )


    # Hidden neuron 1
    parameters['W1'][0][0] = parameters['W1'][0][0] + (
        0.001 * 2 * (y - y_hat) * parameters['W2'][0][0] * X[0][0]
    )

    parameters['W1'][0][1] = parameters['W1'][0][1] + (
        0.001 * 2 * (y - y_hat) * parameters['W2'][0][0] * X[1][0]
    )

    parameters['b1'][0][0] = parameters['b1'][0][0] + (
        0.001 * 2 * (y - y_hat) * parameters['W2'][0][0]
    )


    # Hidden neuron 2
    parameters['W1'][1][0] = parameters['W1'][1][0] + (
        0.001 * 2 * (y - y_hat) * parameters['W2'][0][1] * X[0][0]
    )

    parameters['W1'][1][1] = parameters['W1'][1][1] + (
        0.001 * 2 * (y - y_hat) * parameters['W2'][0][1] * X[1][0]
    )

    parameters['b1'][1][0] = parameters['b1'][1][0] + (
        0.001 * 2 * (y - y_hat) * parameters['W2'][0][1]
    )

    return parameters
```

```python
# epochs implementation

parameters = initialize_parameters([2, 2, 1])

epochs = 5

for i in range(epochs):

    Loss = []

    for j in range(df.shape[0]):

        X = df[['cgpa', 'profile_score']].values[j].reshape(2, 1)

        y = df[['lpa']].values[j][0]

        # Forward propagation
        y_hat, A1 = L_layer_forward(X, parameters)

        y_hat = y_hat[0][0]

        # Update parameters
        update_parameters(parameters, y, y_hat, A1, X)

        # Store loss
        Loss.append((y - y_hat) ** 2)

    # Print loss after completing one epoch
    print("Epoch-" + str(i + 1) + ", Loss: " + str(np.array(Loss).mean()))

parameters
```
